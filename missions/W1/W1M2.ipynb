INSERT에는 두가지 방법이 있음

  1. 특정 컬럼명을 정하고 해당 컬럼에 값 넣기                         
    - 그 외 칼럼은 None으로 채워짐                  
    - 만약 테이블 생성 시 null값을 허용하지 않은 컬럼이 있다면,
      이 방법을 썼을 때 오류가 날 수 있음                 
    - INSERT INTO table_name (col1, col2 ...) VALUES (value1, value2 ...);

  2. 모든 컬럼에 값 넣기
    - 특정 컬럼명을 정하지 않고 모든 컬럼에 값을 넣는 방법             
    - 단, 컬럼의 순서를 테이블의 순서와 일치시켜야 함          
    - INSERT INTO table_name VALUES(value1, value2...);

# SQL JOIN 정리 (MySQL 기준)

## 1. INNER JOIN
두 테이블에 모두 존재하는 행만 조회 (교집합)

```sql
SELECT *
FROM A
INNER JOIN B ON A.id = B.id;
```

- A와 B 양쪽에 매칭되는 데이터만 반환
- 가장 기본적이고 많이 사용됨
- MySQL 지원: 가능

---

## 2. LEFT JOIN (LEFT OUTER JOIN)
왼쪽 테이블(A)은 전부, 오른쪽 테이블(B)은 매칭되는 것만 조회

```sql
SELECT *
FROM A
LEFT JOIN B ON A.id = B.id;
```

- B에 매칭되는 행이 없으면 B 컬럼은 NULL
- A를 기준으로 데이터를 보고 싶을 때 사용
- MySQL 지원: 가능

---

## 3. RIGHT JOIN (RIGHT OUTER JOIN)
오른쪽 테이블(B)은 전부, 왼쪽 테이블(A)은 매칭되는 것만 조회

```sql
SELECT *
FROM A
RIGHT JOIN B ON A.id = B.id;
```

- A에 매칭되는 행이 없으면 A 컬럼은 NULL
- LEFT JOIN과 의미는 같고 방향만 반대
- MySQL 지원: 가능

---

## 4. FULL OUTER JOIN
왼쪽 테이블과 오른쪽 테이블의 모든 행을 조회 (합집합)

```sql
SELECT *
FROM A
FULL OUTER JOIN B ON A.id = B.id;
```

- MySQL 직접 지원: 불가능

### MySQL에서의 대안 (UNION 사용)

```sql
SELECT *
FROM A
LEFT JOIN B ON A.id = B.id
UNION
SELECT *
FROM A
RIGHT JOIN B ON A.id = B.id;
```

- LEFT JOIN 결과 + RIGHT JOIN 결과를 합쳐 FULL JOIN 효과 구현
- UNION은 중복 제거
- 중복 허용 시 UNION ALL 사용

---

## 5. SELF JOIN
같은 테이블을 자기 자신과 JOIN

```sql
SELECT e.name AS employee, m.name AS manager
FROM Employees e
LEFT JOIN Employees m ON e.manager_id = m.id;
```

- 계층 구조 데이터(사원-상사, 댓글-부모 댓글 등)에 사용
- 테이블 별칭(alias)이 필수
- MySQL 지원: 가능

---

## 핵심 요약

- INNER JOIN: 양쪽에 모두 존재하는 데이터
- LEFT JOIN: 왼쪽 기준, 없으면 NULL
- RIGHT JOIN: 오른쪽 기준, 없으면 NULL
- FULL JOIN: 양쪽 전부, MySQL은 UNION으로 구현
- SELF JOIN: 자기 자신과 JOIN

# EXISTS / ANY / ALL

## EXISTS
서브쿼리 결과가 하나라도 존재하면 TRUE

```sql
SELECT *
FROM Products p
WHERE EXISTS (
    SELECT 1
    FROM Customers c
    WHERE c.CustomerID = p.SupplierID
      AND c.Country = 'Mexico'
);
````

* 서브쿼리의 실제 값은 중요하지 않음
* 조건을 만족하는 행이 존재하는지만 판단
* MySQL / SQLite / PostgreSQL 지원

---

## ANY

서브쿼리 결과 중 하나라도 조건을 만족하면 TRUE

```sql
SELECT *
FROM Products
WHERE Price > ANY (
    SELECT Price
    FROM Products
    WHERE CategoryID = 1
);
```

* 비교 연산자와 반드시 함께 사용
* 의미상 IN과 유사한 경우 많음
* MySQL / PostgreSQL 지원
* SQLite 미지원

---

## ALL

서브쿼리 결과의 모든 값에 대해 조건을 만족해야 TRUE

```sql
SELECT *
FROM Products
WHERE Price > ALL (
    SELECT Price
    FROM Products
    WHERE CategoryID = 1
);
```

* 모든 값과 비교
* 최대값 / 최소값 비교에 활용
* MySQL / PostgreSQL 지원
* SQLite 미지원

---

## 핵심 요약

* EXISTS: 존재 여부 확인
* ANY: 하나라도 만족
* ALL: 전부 만족

# SELECT INTO vs INSERT INTO SELECT

## SELECT INTO
SELECT 결과를 이용해 **새 테이블을 생성**하면서 데이터 삽입

```sql
SELECT *
INTO NewTable
FROM OldTable
WHERE condition;
````

* NewTable이 새로 생성됨
* 기존에 테이블이 있으면 오류
* 컬럼 타입은 생성되지만 PK, FK, INDEX 등 제약조건은 대부분 복사되지 않음

지원 DBMS

* SQL Server: 지원
* PostgreSQL: 지원
* MySQL: 미지원
* SQLite: 미지원

---

## INSERT INTO SELECT

이미 존재하는 테이블에 **SELECT 결과를 삽입**

```sql
INSERT INTO TargetTable (col1, col2)
SELECT col1, col2
FROM SourceTable
WHERE condition;
```

* TargetTable이 미리 존재해야 함
* 컬럼 개수와 타입이 맞아야 함
* 제약조건(PK, NOT NULL 등) 그대로 적용

지원 DBMS

* MySQL: 지원
* SQLite: 지원
* PostgreSQL: 지원
* SQL Server: 지원

---

## 비교 표

| 구분        | SELECT INTO | INSERT INTO SELECT |
| --------- | ----------- | ------------------ |
| 테이블 생성    | O           | X                  |
| 기존 테이블 필요 | X           | O                  |
| 제약조건 유지   | X           | O                  |
| MySQL 지원  | X           | O                  |
| SQLite 지원 | X           | O                  |
| 실무 사용 빈도  | 낮음          | 매우 높음              |

---

## 핵심 요약

* SELECT INTO: 조회 결과로 새 테이블 생성
* INSERT INTO SELECT: 기존 테이블에 조회 결과 삽입

NULL Functions (MySQL 기준)

## 1. NULL 기본 개념
- NULL은 값이 아님
- =, != 로 비교 불가
- 반드시 IS NULL / IS NOT NULL 사용

```sql
WHERE PostalCode IS NULL
````

---

## 2. IFNULL

NULL이면 대체값 반환

```sql
SELECT IFNULL(PostalCode, 'N/A')
FROM Customers;
```

* 첫 번째 값이 NULL이면 두 번째 값 반환
* MySQL 지원

---

## 3. COALESCE

여러 값 중 NULL이 아닌 첫 번째 값 반환

```sql
SELECT COALESCE(Phone, Fax, 'N/A')
FROM Suppliers;
```

* IFNULL의 확장 형태
* 인자 개수 제한 없음
* MySQL 지원

---

## 4. NULLIF

두 값이 같으면 NULL 반환, 다르면 첫 번째 값 반환

```sql
SELECT NULLIF(Price, 0)
FROM Products;
```

* 0으로 나누는 오류 방지 등에 사용
* MySQL 지원

---

## 5. CASE (NULL 처리 포함)

조건에 따라 값 분기

```sql
SELECT
CASE
    WHEN PostalCode IS NULL THEN 'Unknown'
    ELSE PostalCode
END
FROM Customers;
```

* 가장 범용적인 NULL 처리 방법
* MySQL 지원

---

## 6. MySQL NULL 함수 지원 요약

| 함수       | MySQL 지원 여부 | 비고          |
| -------- | ----------- | ----------- |
| IFNULL   | 지원          | MySQL 전용 느낌 |
| COALESCE | 지원          | 표준 SQL      |
| NULLIF   | 지원          | 표준 SQL      |
| NVL      | 미지원         | Oracle 전용   |
| CASE     | 지원          | 가장 범용       |

---

## 핵심 요약

* NULL 비교: IS NULL
* 값 대체: IFNULL, COALESCE
* 조건 처리: CASE
* 안전 처리: NULLIF

# Stored Procedure (프로시저) 개념 정리  
(W3Schools SQL Tutorial 기준 요약)

## Stored Procedure란?
Stored Procedure는 **데이터베이스에 저장된 SQL 코드 묶음**이다.  
자주 사용하는 SQL 문을 하나의 이름으로 저장해 두고, 필요할 때 호출해서 실행한다.

```sql
CALL procedure_name();
````

---

## 왜 사용하는가?

W3Schools 기준으로 프로시저의 목적은 다음과 같다.

* 동일한 SQL을 반복해서 작성하지 않아도 됨
* 복잡한 SQL 로직을 하나의 단위로 관리
* 네트워크 트래픽 감소 (SQL을 매번 전송하지 않음)
* 보안 향상 (테이블 직접 접근 대신 프로시저만 실행)

---

## 기본 문법 (MySQL / SQL Server)

### 프로시저 생성

```sql
CREATE PROCEDURE SelectAllCustomers
AS
SELECT * FROM Customers;
```

### 프로시저 실행

```sql
EXEC SelectAllCustomers;
```

---

## 파라미터가 있는 프로시저

```sql
CREATE PROCEDURE SelectCustomersByCountry
    @Country VARCHAR(50)
AS
SELECT *
FROM Customers
WHERE Country = @Country;
```

실행:

```sql
EXEC SelectCustomersByCountry 'Mexico';
```

* 입력 파라미터를 받아 동적인 쿼리 실행 가능
* 조건에 따라 다른 결과 반환 가능

---

## 프로시저의 특징 요약

* DB 내부에 저장됨
* SELECT, INSERT, UPDATE, DELETE 모두 포함 가능
* 조건문(IF), 반복문(LOOP)을 포함할 수 있음 (DBMS에 따라 차이)
* 트랜잭션 처리 가능
